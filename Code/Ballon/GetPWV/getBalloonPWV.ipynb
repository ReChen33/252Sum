```python
# save a csv file of upper air data from Wyoming for a given date and station
import os
from siphon.simplewebservice.wyoming import WyomingUpperAir
from datetime import datetime
import time

#base_dir = os.path.dirname(os.path.abspath(__file__))

data_balloon_list = []

years = [2024,2025]
for month in range(1,13):
    for days in range(1,32):
        try:
            data00_2024 = datetime(years[0], month, days, 0)
            data12_2024 = datetime(years[0], month, days, 12)

            data_balloon_list.append(data00_2024)
            data_balloon_list.append(data12_2024)

            if month == 1 or month == 2:
                data00_2025 = datetime(years[1], month, days, 0)
                data12_2025 = datetime(years[1], month, days, 12)
                data_balloon_list.append(data00_2025)
                data_balloon_list.append(data12_2025)

        except:
            print(f'Invalid date: {years[0]}-{month}-{days}\n')
            continue

data_balloon_list.sort()
print(f'Total valid dates to process: {len(data_balloon_list)}\n\n')

csv_fn = 'Alldate_balloon_SP.csv'
with open(csv_fn, 'w+') as f:
    f.write('Date(YYYYMMDD_HHMMSS),PWV(um)\n')

for date in data_balloon_list:
    #print(f'Processing date: {date}\n')
    station = '89009' #South Pole Station
    try:
        time.sleep(0.1)
        df = WyomingUpperAir.request_data(date, station)        
    except:
        print(f'No data for {station} on {date}')
        continue

    pwv = df['pw'].values[0]*1000  # convert from mm to um
    #save pwv to a csv file
    with open(csv_fn, 'a+') as f:
        f.write(f'{date.strftime("%Y%m%d_%H%M%S")},{pwv:.3f}\n')
    

```
