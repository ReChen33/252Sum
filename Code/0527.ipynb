```python
import numpy as np
import matplotlib.pyplot as plt

plt.subplot(211)
        
plt.plot(np.random.randn(1000))
plt.title('Random Data Plot')
plt.xlabel('Index')
plt.ylabel('Value')
plt.grid(True)

plt.subplot(221)
plt.plot(np.random.randn(1000).cumsum())
plt.title('Cumulative Sum of Random Data')
plt.xlabel('Index')
plt.ylabel('Cumulative Value')
plt.grid(True)

plt.subplot(212)
plt.plot(np.random.randn(1000).cumsum(), label='Cumulative Sum')
plt.plot(np.random.randn(1000).cumsum() + 10, label='Cumulative Sum + 10')
plt.title('Multiple Cumulative Sums')
plt.xlabel('Index')
plt.ylabel('Cumulative Value')
plt.xlim(0, 1000)
plt.legend()
plt.grid(True)
plt.close
```
