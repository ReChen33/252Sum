```python
# read npy files 

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

npy_data = np.load('ProTest_time.npy')
plt.hist(npy_data[:-1], bins=100)
plt.xlabel('Time taken (seconds)')
plt.ylabel('Number of occurrences')
plt.title('Histogram of Time Taken for npy')
plt.grid(True)
plt.yscale('log')
plt.show()

csv_data = pd.read_csv('ProTest_time.csv')
plt.hist(csv_data['Time Taken (s)'], bins=100)
plt.xlabel('Time taken (seconds)')
plt.ylabel('Number of occurrences')
plt.title('Histogram of Time Taken for csv')
plt.grid(True)
plt.yscale('log')
plt.show()
```
